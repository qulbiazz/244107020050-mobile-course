1.	Implementasikan inverse citra pada Google Colaboratory menggunakan formula yang terdapat pada bagian Ulasan Teori, sehingga menghasilkan keluaran seperti berikut

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import drive
print('Operasi Inverse Citra (Citra Negatif)')
print('--------------------------------')


#sesuaikan dengan file dan folder di drive anda
original = cv.imread('/content/drive/MyDrive/PCVK/peppers.jpg')
inverse_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      inverse_image[y, x, c] = 255 - original[y, x, c]

#cara simple tanpa for loop
#brightness_images = cv.convertScaleAbs(original, beta=brightness)

final_frame=cv.hconcat((original, inverse_image))
cv2_imshow(final_frame)
```

2.	Implementasikan transformasi contrast pada Google Colaboratory menggunakan formula yang terdapat pada bagian Ulasan Teori untuk kontras, sehingga menghasilkan keluaran seperti berikut:

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import drive
print('Mengubah kontras dan tingkat kecerahan citra')
print('--------------------------------')
try:
  brightness = int(input('Masukkan tingkat kecerahan: '))
  contrast = float(input('Masukkan kontras: '))
except ValueError:
    print('Error, not a number')
    brightness = 0
    contrast = 1.0

#sesuaikan dengan file dan folder di drive anda
original = cv.imread('/content/drive/MyDrive/PCVK/sunset.jpg')
contrast_image = np.zeros(original.shape, original.dtype)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      nilai_baru = contrast * original[y, x, c] + brightness
      contrast_image[y, x, c] = np.clip(nilai_baru, 0, 255)

#cara simple tanpa for loop
#brightness_images = cv.convertScaleAbs(original, beta=brightness)

final_frame=cv.hconcat((original, contrast_image))
cv2_imshow(final_frame)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import drive
print('Mengubah tingkat kecerahan citra dengan Transformasi Log')
print('--------------------------------')
try:
  brightness = int(input('Masukkan nilai kecerahan: '))
except ValueError:
    print('Error, not a number')
    brightness = 0

#sesuaikan dengan file dan folder di drive anda
original = cv.imread('/content/drive/MyDrive/PCVK/houses.jpg')
logarithmBrightness_image = np.zeros(original.shape, original.dtype)

# 1. Hitung konstanta c agar rentang hasil berada di 0 - 255
    # c = 255 / log(1 + 255)
c_const = 255.0 / np.log(1.0 + 255.0)

#akses per piksel
#np.clip digunakan untuk melakukan truncate pixel (nilai dibatasi antara 0-255)
for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for ch in range(original.shape[2]):
                r = float(original[y, x, ch])
                s = c_const * np.log(1.0 + r)
                logarithmBrightness_image[y, x, ch] = np.clip(s, 0, 255)

#cara simple tanpa for loop
#brightness_images = cv.convertScaleAbs(original, beta=brightness)

final_frame=cv.hconcat((original, logarithmBrightness_image))
cv2_imshow(final_frame)
```

4.	Implementasikan transformasi grayscale menggunakan metode averaging, lightness, dan luminance pada Google Colaboratory menggunakan formula yang terdapat pada bagian Ulasan Teori, sehingga menghasilkan keluaran seperti berikut:

a.	Averaging

```python
import matplotlib.pyplot as plt
# 1. Baca citra dan ubah format ke RGB
path = '/content/drive/MyDrive/PCVK/peppers.jpg'
img_bgr = cv.imread(path)
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
h, w = img_rgb.shape[:2]

# Inisialisasi matriks 2D untuk masing-masing metode
gray_avg = np.zeros((h, w), dtype=np.uint8)

# 2. Akses per piksel menerapkan rumus masing-masing
for y in range(h):
    for x in range(w):
        r = float(img_rgb[y, x, 0])
        g = float(img_rgb[y, x, 1])
        b = float(img_rgb[y, x, 2])

        # a. Averaging: (R + G + B) / 3
        gray_avg[y, x] = np.clip((r + g + b) / 3.0, 0, 255)

# a. Averaging
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.subplot(1, 2, 2)
plt.imshow(gray_avg, cmap='gray')
plt.suptitle('a. Averaging', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

b.	Lightness

```python
import matplotlib.pyplot as plt
# 1. Baca citra dan ubah format ke RGB
path = '/content/drive/MyDrive/PCVK/peppers.jpg'
img_bgr = cv.imread(path)
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
h, w = img_rgb.shape[:2]

# Inisialisasi matriks 2D untuk masing-masing metode
gray_lightness = np.zeros((h, w), dtype=np.uint8)

# 2. Akses per piksel menerapkan rumus masing-masing
for y in range(h):
    for x in range(w):
        r = float(img_rgb[y, x, 0])
        g = float(img_rgb[y, x, 1])
        b = float(img_rgb[y, x, 2])

        # b. Lightness: (max(R, G, B) + min(R, G, B)) / 2
        gray_lightness[y, x] = np.clip((max(r, g, b) + min(r, g, b)) / 2.0, 0, 255)

# b. Lightness
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.subplot(1, 2, 2)
plt.imshow(gray_lightness, cmap='gray')
plt.suptitle('b. Lightness', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

c.	Luminance

```python
import matplotlib.pyplot as plt
# 1. Baca citra dan ubah format ke RGB
path = '/content/drive/MyDrive/PCVK/peppers.jpg'
img_bgr = cv.imread(path)
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
h, w = img_rgb.shape[:2]

# Inisialisasi matriks 2D untuk masing-masing metode
gray_luminance = np.zeros((h, w), dtype=np.uint8)

# 2. Akses per piksel menerapkan rumus masing-masing
for y in range(h):
    for x in range(w):
        r = float(img_rgb[y, x, 0])
        g = float(img_rgb[y, x, 1])
        b = float(img_rgb[y, x, 2])

        # c. Luminance: 0.21*R + 0.72*G + 0.07*B
        gray_luminance[y, x] = np.clip(0.21 * r + 0.72 * g + 0.07 * b, 0, 255)

# c. Luminance
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.subplot(1, 2, 2)
plt.imshow(gray_luminance, cmap='gray')
plt.suptitle('c. Luminance', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

5.	Tampilkanlah warna tertentu pada citra, dan ubah warna lain menjadi grayscale. Misal, tampilkan warna merah pada citra masukan dan ubah bagian lain yang tidak berwarna merah menjadi grayscale seperti pada contoh berikut:

```python
import matplotlib.pyplot as plt

# 1. Baca citra dan konversi ruang warna
path = '/content/drive/MyDrive/PCVK/peppers.jpg'
img_bgr = cv.imread(path)
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

# 2. Buat versi grayscale standar menggunakan OpenCV
img_gray = cv.cvtColor(img_bgr, cv.COLOR_BGR2GRAY)
# Ubah dimensi grayscale 1-channel ke 3-channel (H, W, 3)
gray_3ch = cv.cvtColor(img_gray, cv.COLOR_GRAY2RGB)

# 3. Buat mask untuk mendeteksi warna merah (Red > 130 dan dominan terhadap G & B)
r = img_rgb[:, :, 0]
g = img_rgb[:, :, 1]
b = img_rgb[:, :, 2]

mask_merah = (r > 130) & (r > g * 1.4) & (r > b * 1.4)

# 4. Gabungkan: piksel merah mengambil citra asli, sisanya mengambil citra abu-abu
hasil = np.where(mask_merah[:, :,  ], img_rgb, gray_3ch)

# 5. Tampilkan hasil
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Citra Masukan")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(hasil)
plt.title("Hanya Warna Merah")
plt.axis('off')

plt.tight_layout()
plt.show()
```
