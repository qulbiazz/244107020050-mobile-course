Tuliskan pertanyaan Anda, rancangan program, data hasilnya, dan kesimpulan Anda.

Pertanyaan
- Berapa persentase area/piksel berpenetrasi gelap (intensitas piksel rata-rata $< 100$) pada foto KTP/KTM, dan bagaimana distribusi dominasi warna (Red, Green, Blue) rata-ratanya?

Rancangan Program
- Membaca citra menggunakan OpenCV (cv.imread) lalu mengubah representasi warna ke RGB (cv.COLOR_BGR2RGB).
- Menghitung rata-rata intensitas per piksel di ketiga kanal $\frac(R + G + B)/3 sebagai representasi tingkat kecerahan (brightness).
- Melakukan thresholding boolean untuk menghitung rasio jumlah piksel bernilai $< 100$ terhadap total seluruh piksel citra.
- Menghitung nilai rata-rata tiap kanal individual (R, G, dan B) untuk melihat kecenderungan warna latar/dokumen.

```python
import cv2 as cv
import numpy as np

# 1. Baca citra dan ubah format BGR ke RGB
img = cv.imread('Ktp.jpg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

# 2. Hitung intensitas rata-rata per piksel
intensitas = np.mean(img_rgb, axis=2)

# 3. Hitung persentase piksel gelap (< 100)
total_piksel = intensitas.size
piksel_gelap = np.sum(intensitas < 100)
persentase_gelap = (piksel_gelap / total_piksel) * 100

# 4. Hitung rata-rata tiap kanal warna
avg_r = np.mean(img_rgb[:, :, 0])
avg_g = np.mean(img_rgb[:, :, 1])
avg_b = np.mean(img_rgb[:, :, 2])

# Tampilkan data hasil perhitungan
print("=== HASIL PENGUKURAN CITRA ===")
print(f"Dimensi citra (Shape)       : {img_rgb.shape}")
print(f"Total seluruh piksel        : {total_piksel:,} piksel")
print(f"Jumlah piksel gelap (< 100) : {piksel_gelap:,} piksel")
print(f"Persentase piksel gelap     : {persentase_gelap:.2f} %")
print(f"Rata-rata Intensitas Red    : {avg_r:.2f}")
print(f"Rata-rata Intensitas Green  : {avg_g:.2f}")
print(f"Rata-rata Intensitas Blue   : {avg_b:.2f}")
```

TUGAS PRAKTIKUM - Aplikasi

Pada bagian ini, Anda hanya diperbolehkan memakai fungsi yang telah dipelajari pada modul ini: pembacaan berkas, pengirisan, cv2.resize, cv2.flip, cv2.cvtColor, fungsi penggambaran OpenCV, serta pembuatan kanvas dengan NumPy. Selain itu, program harus bersifat adaptif, dimana setiap ukuran, posisi, atau ketebalan harus dihitung dari ukuran citra masukan, bukan ditulis sebagai angka tetap pada kode program.

Studi Kasus 1: Panitia sebuah kegiatan perlu membubuhkan label identitas pada ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel, sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga proporsinya tetap sama.

Studi Kasus 2: Marketplace seperti Tokopedia dan Shopee menampilkan foto produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja. Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna. Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas, tambahkan logo nama toko berukuran proporsional pada gambar

```python
import cv2 as cv
import matplotlib.pyplot as plt

# 1. Fungsi penambahan label identitas adaptif
def tambah_label_adaptif(img, teks="Dokumentasi PCVK - 2026"):
    output = img.copy()
    h, w = output.shape[:2]

    # Rumus adaptif dari soal: bar = tinggi // 12
    tinggi_bar = h // 12
    font_scale = (tinggi_bar / 50) * 0.55
    thickness = max(1, int(tinggi_bar / 30))

    # Buat bar hitam solid di bagian bawah
    cv.rectangle(output, pt1=(0, h - tinggi_bar), pt2=(w, h), color=(0, 0, 0), thickness=-1)

    # Letakkan teks di sisi kiri bawah di dalam bar
    posisi_x = int(w * 0.02)
    posisi_y = int(h - (tinggi_bar * 0.35))
    cv.putText(output, teks, (posisi_x, posisi_y), cv.FONT_HERSHEY_SIMPLEX,
               font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return output, tinggi_bar

# 2. Baca kedua citra pemandangan (lanskap & potret)
# Sesuaikan path/nama file dengan file yang telah diunggah di Colab Anda
path_lanskap = '/content/drive/MyDrive/PCVK/pemandangan_lanskap.jpg'  # foto pemandangan horizontal
path_potret  = '/content/drive/MyDrive/PCVK/pemandangan_potret.jpg'    # foto pemandangan vertikal

img_lanskap = cv.imread(path_lanskap)
img_potret = cv.imread(path_potret)

# Ubah ke RGB
img_lanskap = cv.cvtColor(img_lanskap, cv.COLOR_BGR2RGB)
img_potret = cv.cvtColor(img_potret, cv.COLOR_BGR2RGB)

# 3. Sesuaikan ukuran masukan sesuai gambar acuan (900x600 dan 480x760)
img1 = cv.resize(img_lanskap, (900, 600))
img2 = cv.resize(img_potret, (480, 760))

# 4. Beri label adaptif pada kedua citra
hasil1, bar1 = tambah_label_adaptif(img1)
hasil2, bar2 = tambah_label_adaptif(img2)

# 5. Tampilkan dalam susunan 2x2 seperti pada gambar acuan
plt.figure(figsize=(11, 10))

# Baris 1: Citra Lanskap
plt.subplot(2, 2, 1)
plt.imshow(img1)
plt.title(f"Masukan {img1.shape[1]} x {img1.shape[0]} px")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(hasil1)
plt.title(f"Keluaran: bar = {img1.shape[0]}//12 = {bar1} px")
plt.axis('off')

# Baris 2: Citra Potret
plt.subplot(2, 2, 3)
plt.imshow(img2)
plt.title(f"Masukan {img2.shape[1]} x {img2.shape[0]} px")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(hasil2)
plt.title(f"Keluaran: bar = {img2.shape[0]}//12 = {bar2} px")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

def buat_thumbnail_persegi(img, ukuran_target=600, warna_bg=(240, 240, 240)):
    h_asli, w_asli = img.shape[:2]

    # 1. Hitung rasio adaptif
    skala = min(ukuran_target / w_asli, ukuran_target / h_asli)
    w_baru = int(w_asli * skala)
    h_baru = int(h_asli * skala)

    img_resized = cv.resize(img, (w_baru, h_baru), interpolation=cv.INTER_AREA)

    # 2. Buat kanvas persegi 1:1
    kanvas = np.full((ukuran_target, ukuran_target, 3), warna_bg, dtype=np.uint8)

    # 3. Tempel tepat di tengah kanvas
    x_offset = (ukuran_target - w_baru) // 2
    y_offset = (ukuran_target - h_baru) // 2
    kanvas[y_offset:y_offset + h_baru, x_offset:x_offset + w_baru] = img_resized

    return kanvas

# Baca citra lanskap dari Google Drive
path_lanskap = '/content/drive/MyDrive/PCVK/pemandangan_lanskap.jpg'
img_lanskap = cv.imread(path_lanskap)
img_lanskap_rgb = cv.cvtColor(img_lanskap, cv.COLOR_BGR2RGB)

# Buat thumbnail persegi 1:1
hasil_thumbnail = buat_thumbnail_persegi(img_lanskap_rgb, ukuran_target=600)

# Tampilkan gambar bersih tanpa judul teks/keterangan
plt.figure(figsize=(6, 6))
plt.imshow(hasil_thumbnail)
plt.axis('off')
plt.show()
```
